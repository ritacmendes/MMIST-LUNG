```python
# Function to check if a column is binary (YES/NO, True/False, 1/0, Y/N, T/F)
def is_binary(col):
    unique_vals = set(col.dropna().unique())
    binary_sets = [
        {'YES', 'NO'}, {'Yes','No'}, {'yes','no'},
        {True, False}, {1,0},
    ]
    return any(unique_vals.issubset(s) for s in binary_sets)

for c in df.columns:
    if is_binary(df[c]):
        df[c] = df[c].map(lambda x: 1 if str(x).upper() in ['YES','Y','TRUE','T','1'] else 0)

def drop_constant_columns(df):
    """
    Remove columns where all non-NaN entries have the same value.
    """
    return df.loc[:, df.nunique(dropna=True) > 1]
```

# DEMOGRAPHICS MAPPINGS

```python
dem_cols = [
       'cases.submitter_id', 
    'demographic.days_to_birth',
       'demographic.days_to_death', 
       'demographic.ethnicity', 'demographic.race',
       'demographic.sex_at_birth',
       'demographic.vital_status',
        'exposures.pack_years_smoked',
    'exposures.tobacco_smoking_onset_year',
    'exposures.tobacco_smoking_status',
       'exposures.tobacco_smoking_quit_year',
]

dem_df = df[dem_cols]

gender_map = {
    'MALE':0,
    'FEMALE':1
}

ethnicity_map = {
    'NOT HISPANIC OR LATINO':0,
    'HISPANIC OR LATINO':1
}

race_map = {
    'WHITE':0,
    'ASIAN':1,
    'BLACK OR AFRICAN AMERICAN':2,
       'AMERICAN INDIAN OR ALASKA NATIVE':3
}

tobacco_smoking_status_map = {
    'LIFELONG NON-SMOKER':0,
    'CURRENT REFORMED SMOKER FOR > 15 YRS':1,
    'CURRENT REFORMED SMOKER FOR < OR = 15 YRS':2,
    'CURRENT SMOKER':3,    
}

dem_df['exposures.tobacco_smoking_onset_year'] = pd.to_numeric(
    dem_df['exposures.tobacco_smoking_onset_year'],
    errors='coerce'
)

dem_df['exposures.tobacco_smoking_quit_year'] = pd.to_numeric(
    dem_df['exposures.tobacco_smoking_quit_year'],
    errors='coerce'
)
dem_df['demographic.days_to_birth'] = pd.to_numeric(
    dem_df['demographic.days_to_birth'],
    errors='coerce'
)
dem_df['demographic.gender'] = dem_df['demographic.sex_at_birth'].map(gender_map)
dem_df['demographic.ethnicity'] = dem_df['demographic.ethnicity'].map(ethnicity_map)
dem_df['demographic.race'] = dem_df['demographic.race'].map(race_map)
dem_df['exposures.tobacco_years_smoked']=dem_df['exposures.tobacco_smoking_quit_year']-dem_df['exposures.tobacco_smoking_onset_year']
dem_df['exposures.tobacco_smoking_status'] = dem_df['exposures.tobacco_smoking_status'].map(tobacco_smoking_status_map)

# 1) convert days to age in years
age_years = -dem_df['demographic.days_to_birth'] / 365.25

# 2) create 10-year bins (0–9 → 0, 10–19 → 1, etc.)
dem_df['diagnoses.age_bin'] = (age_years // 10).astype('Int64')

death_days = pd.to_numeric(dem_df['demographic.days_to_death'], errors='coerce')

dem_df['vital_status_12'] = np.where(
    (death_days <= 365) | (death_days.isna()),
    1,
    0
)
dem_df=dem_df.drop(columns=['demographic.sex_at_birth','demographic.vital_status','demographic.days_to_birth', 'exposures.tobacco_smoking_quit_year', 'exposures.tobacco_smoking_onset_year'])

```

# TREATMENT MAPPINGS

```python
def classify_treatment(t):
    if pd.isna(t):
        return np.nan
    
    t = t.lower()
    
    if "radiation" in t:
        return "Radiation"
    elif "chemotherapy" in t:
        return "Chemotherapy"
    elif "pharmaceutical" in t:
        return "Pharmaceutical"
    elif "surgery" in t:
        return "Surgery"
    else:
        return np.nan
        
cols_to_drop = ['cohort','cases.case_id',
                'follow_ups.follow_up_id',
                'diagnoses.age_at_diagnosis',
               #'diagnoses.classification_of_tumor', # SO SELECIONEI PRIMARY,
                'diagnoses.primary_diagnosis', # TEM ESTA INFORMACAO NAS OUTRAS COLUNAS
                'diagnoses.laterality', # TEM ESTA INFORMAÇÃO NO ICD10
               ]

treatment_cols = ['cases.submitter_id','diagnoses.days_to_diagnosis','diagnoses.classification_of_tumor','treatments.days_to_treatment_start','treatments.days_to_treatment_end', 'treatments.treatment_type', 'treatments.treatment_outcome']

treatment_outcome_map = {
    'Treatment Ongoing': np.nan,  
    #'Treatment Ongoing': 0,
    'Stable Disease': 1,
    'Progressive Disease': 2,
    'Partial Response' : 3,
    'Complete Response' : 4,    
}

df = df.drop(columns=cols_to_drop, errors='ignore')
# encode binary to 0/1
for c in df.columns:
    if is_binary(df[c]):
        df[c] = df[c].map(lambda x: 1 if str(x).upper() in ['YES','Y','TRUE','T','1'] else 0)

df['treatments.treatment_outcome'] = df['treatments.treatment_outcome'].map(treatment_outcome_map)
df = df[treatment_cols]
df=df.dropna(subset='treatments.treatment_type')
df = df[
    df['treatments.treatment_type'].fillna('').str.startswith(
        ('Chemo', 'Radiation', 'Surgery', 'Pharm')
    )
].copy()


df['treatments.days_to_treatment_start'] = (
    df['treatments.days_to_treatment_start']
    .fillna(df['diagnoses.days_to_diagnosis'])
)
df['treatments.treatment_duration']=df['treatments.days_to_treatment_end']-df['treatments.days_to_treatment_start']
df = df.drop(columns='treatments.days_to_treatment_end')
```

```python
df = df.drop_duplicates()
```

```python
patient_df = pd.read_csv('features/TCGA-encoded/new-demographics-encoded.csv')
treatment_overview_df = pd.read_csv('features/TCGA/new-treatment-primaryTumor_overview.csv')
# drop entries where t_start is <0 (usually related to pre-lung cancer surgery events)
treatment_overview_df = treatment_overview_df[
    (treatment_overview_df["treatments.days_to_treatment_start"] >= 0) |
    (treatment_overview_df["treatments.days_to_treatment_start"].isna())
]
print(treatment_overview_df["treatments.days_to_treatment_start"].isna().mean() * 100)
```

```python
# Standardize treatment type
def map_type(x):
    x = x.upper()
    if 'RADIATION' in x:
        return 'RADIATION'
    elif 'PHARM' in x or 'CHEMO' in x or 'HORMONE' in x:
        return 'PHARM'
    elif 'SURGERY' in x:
        return 'SURGERY'
    else:
        return x  # keep as-is if unknown

treatment_overview_df['treatment_type_std'] = treatment_overview_df['treatments.treatment_type'].apply(map_type)

# Assume death_days column exists for each patient
patient_death = patient_df.set_index('cases.submitter_id')['demographic.days_to_death']  # or events["days_to_death"]

## Fill remaining NaNs with default values
treatment_overview_df['treatments.days_to_treatment_start'] = treatment_overview_df['treatments.days_to_treatment_start'].fillna(
    treatment_overview_df['treatment_type_std'].map(defaults)
)
# Clip treatments so they happen **before death**
def clip_to_death(row):
    pid = row['cases.submitter_id']
    death_day = patient_death.get(pid, None)
    t_start = row['treatments.days_to_treatment_start']
    if death_day is not None and death_day >= 0:
        return min(t_start, death_day)  # can't start after death
    return t_start

treatment_overview_df['treatments.days_to_treatment_start'] = treatment_overview_df.apply(clip_to_death, axis=1)

# Check the percentage of NaNs left
nan_pct = treatment_overview_df['treatments.days_to_treatment_start'].isna().mean() * 100
print(f"Remaining NaNs: {nan_pct:.2f}%")

# Optional: drop the helper column if no longer needed
treatment_overview_df.drop(columns=['treatment_type_std'], inplace=True)
```

```python
treatment_overview_df['treatments.treatment_duration_days'] = treatment_overview_df['treatments.days_to_treatment_end'] - treatment_overview_df['treatments.days_to_treatment_start']
# Replace negative durations with NaN
treatment_overview_df.loc[
    treatment_overview_df['treatments.treatment_duration_days'] < 0,
    'treatments.treatment_duration_days'
] = np.nan
```

```python
locoregional_sites = {
    "LUNG",
    "LYMPH NODE(S) MEDIASTINAL",
    "CHEST WALL",
    "LOCOREGIONAL SITE"
}

distant_sites = {
    "DISTANT SITE",
    "BREAST",
    "THYROID",
    "VULVA",
    "VAGINA",
    "SKIN",
    "COLON",
    "PROSTATE",
    "EPIGLOTTIS"
}

# Only surgery rows
surgery_df = treatment_overview_df[treatment_overview_df["treatments.treatment_type"] == "SURGERY, NOS"].copy()

# Explode multi-site strings
surgery_df = surgery_df.assign(
    site = surgery_df["treatments.treatment_anatomic_sites"]
                .str.split("|")
).explode("site")

# Strip whitespace
surgery_df["site"] = surgery_df["site"].str.strip()

# Map to numeric code
def site_code(site):
    if pd.isna(site):
        return 0
    if site in distant_sites:
        return 2
    if site in locoregional_sites:
        return 1
    return 0

surgery_df["site_code"] = surgery_df["site"].map(site_code)

# Collapse per patient + t_start: take max (DISTANT overrides LOCO)
surgery_codes = (
    surgery_df.groupby(["cases.submitter_id", "treatments.days_to_treatment_start"])["site_code"]
    .max()
    .reset_index(name="surgery")
)

# If a patient-event has no SURGERY,NOS row, it will be missing → NaN
```

```python
# drug vocabulary
drug_vocab = sorted(treatment_overview_df["treatments.therapeutic_agents"].dropna().unique())
drug_to_idx = {d: i for i, d in enumerate(drug_vocab)}
n_drugs = len(drug_vocab)
group_cols = [
    "cases.submitter_id",
    "treatments.days_to_treatment_start"
]
rows = []

for (pid, t_start), group in treatment_overview_df.groupby(group_cols):

    chemo_vec = np.zeros(n_drugs, dtype=int)

    # fill chemo vector
    drugs = group["treatments.therapeutic_agents"].dropna().unique()
    for d in drugs:
        if d in drug_to_idx:
            chemo_vec[drug_to_idx[d]] = 1

    # detect radiation therapy
    radiation = int(
        group["treatments.treatment_type"]
        .str.contains("RADIATION", case=False, na=False)
        .any()
    )
    t_duration = group["treatments.treatment_duration_days"].dropna()
    t_duration = t_duration.max() if not t_duration.empty else np.nan
    rows.append({
        "cases.submitter_id": pid,
        "treatments.days_to_treatment_start": t_start,
        "treatments.treatment_duration_days": t_duration,
        "chemo_vector": chemo_vec,
        "radiation_therapy": radiation
    })

patient_treatment_df = pd.DataFrame(rows)
```

```python
patient_treatment_df = patient_treatment_df.merge(
    surgery_codes,  # DataFrame with columns: cases.submitter_id, t_start, surgery
    on=["cases.submitter_id", "treatments.days_to_treatment_start"],
    how="left"
)
```

```python
# Function to clean string and convert to list of ints
def parse_chemo_vector(s):
    # Remove brackets and newlines, then convert to ints
    clean = s.replace('[','').replace(']','').replace('\n',' ')
    return [int(x) for x in clean.split()]

# Apply to the column
df['chemo_vector'] = df['chemo_vector'].apply(parse_chemo_vector)

print(df['chemo_vector'][0])  # Now it’s a proper Python list
```
